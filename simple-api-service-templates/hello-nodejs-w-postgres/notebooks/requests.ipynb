```python
import requests

BASE_URL = "http://localhost:3002/api"
API_ENDPOINT = f"{BASE_URL}"
```

```python
# GET root
requests.get(f"{BASE_URL}/hello")  # Get all posts
```

```python
# POST root
requests.post(f"{BASE_URL}/hello", {"name": "Node.js w. Express"})  # Create a post
```

```python
# PUT root
requests.put(f"{BASE_URL}/hello/1", {"name": "Node.js w. Express"})  # Update a post
```

```python
# PATCH root
requests.patch(f"{BASE_URL}/hello/1", {"name": "Node.js w. Express"})  # Update a post
```

```python
# DELETE root
requests.delete(f"{BASE_URL}/hello/1")  # Delete a post
```

```python
# Users
requests.get(f"{BASE_URL}/users").json()
```

```python
# Customers and their orders
requests.get(f"{BASE_URL}/customers").json()
requests.get(f"{BASE_URL}/customers/1/orders").json()
```

```python
# Orders with pending totals and line items
requests.get(f"{BASE_URL}/orders").json()
requests.get(f"{BASE_URL}/orders/pending-totals").json()
requests.get(f"{BASE_URL}/orders/1/items").json()
```
